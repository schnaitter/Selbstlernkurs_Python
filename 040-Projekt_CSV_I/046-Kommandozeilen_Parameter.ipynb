# Kommandozeilen-Parameter

Bisher stand der Dateipfad fest im Code. Wollen wir dasselbe Programm auf eine
andere CSV-Datei anwenden, müssten wir den Code ändern. Eleganter ist es, den
Dateipfad beim **Aufruf** des Programms mitzugeben:

```console
$ python3 csv_uebersicht.py daten.csv
```

Solche zusätzlichen Angaben heißen **Kommandozeilen-Parameter** oder
**-argumente** (command-line arguments). In Python stehen sie im Modul `sys`
in der Liste `sys.argv` zur Verfügung.

## `sys.argv`

`sys.argv` ist eine Liste von Strings:

- `sys.argv[0]` ist der Name des Skripts (beziehungsweise des Programms),
- `sys.argv[1]`, `sys.argv[2]`, … sind die übergebenen Parameter.

Bei `python3 csv_uebersicht.py daten.csv` gilt also:

| Index | Wert |
| --- | --- |
| `sys.argv[0]` | `"csv_uebersicht.py"` |
| `sys.argv[1]` | `"daten.csv"` |

````python
import sys

# In einem Notebook enthält `sys.argv` die Startparameter des Kernels.
# Diese haben nichts mit Ihrem Skript zu tun. Wichtig ist:
print("Anzahl der Einträge in sys.argv:", len(sys.argv))
print("Der erste Eintrag ist der Programmname:", sys.argv[0].split("/")[-1])
````

```{note}
In einem Jupyter-Notebook gibt es keine Kommandozeilen-Parameter für das
Notebook selbst. `sys.argv` enthält hier die Startparameter des Kernels, nicht
Ihre eigenen Parameter. Wir setzen die Liste weiter unten zu
Demonstrationszwecken manuell.
```

## Den Dateipfad übergeben

Im {term}`Skript` steht `sys.argv[1]` für den Dateipfad. Fehlt das Argument, geben wir
eine kurze Hilfe aus und beenden das Programm mit einem Fehlercode.

```python
def main():
    if len(sys.argv) != 2:
        print("Aufruf: csv_uebersicht.py CSV-DATEI")
        sys.exit(1)
    uebersicht(sys.argv[1])
```

`sys.exit(1)` beendet das Programm. Der Wert `1` signalisiert einen Fehler –
nützlich, wenn das Skript in anderen Programmen oder in der Shell aufgerufen
wird.

## Das Skript aus dem Notebook aufrufen

Mit dem Modul `subprocess` können wir das Skript so starten, wie wir es auch im
Terminal tun würden.

````python
import subprocess
import sys

ergebnis = subprocess.run(
    [
        sys.executable,
        "csv_uebersicht.py",
        "../assets/data/books_powerlaw_dataset.csv",
    ],
    capture_output=True,
    text=True,
)

print("Rückgabecode:", ergebnis.returncode)
print(ergebnis.stdout)
````

Fehlt der Parameter, erklärt das Skript die korrekte Verwendung:

````python
ergebnis = subprocess.run(
    [sys.executable, "csv_uebersicht.py"],
    capture_output=True,
    text=True,
)

print("Rückgabecode:", ergebnis.returncode)
print(ergebnis.stdout.strip())
````

## Parameter im Notebook simulieren

Da Notebooks keine eigenen Kommandozeilen-Parameter erhalten, setzen wir
`sys.argv` hier manuell. So lässt sich die Logik auch im Notebook testen.

````python
sys.argv = ["csv_uebersicht.py", "../assets/data/books_powerlaw_dataset.csv"]

print("Programmname:", sys.argv[0])
print("Übergebener Dateipfad:", sys.argv[1])
````

## Zusammenfassung

- Kommandozeilen-Parameter stehen in `sys.argv` (Liste von Strings).
- `sys.argv[0]` ist der Programmname, ab `sys.argv[1]` folgen die Parameter.
- Prüfen Sie die Anzahl der Parameter und geben Sie bei fehlenden Argumenten
  eine Hilfe aus (`sys.exit(1)`).
- Über `if __name__ == "__main__": main()` wird das Skript erst beim direkten
  Aufruf ausgeführt.

```{hint}
In professionellen Werkzeugen übernimmt das Modul `argparse` das Auswerten von
Parametern inklusive Hilfe und Optionen. `sys.argv` ist der einfache Einstieg.
```
